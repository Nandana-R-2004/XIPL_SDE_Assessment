````python
# ============================================================
# Q4 — CELL 5
# LOAD OPENCLIP FOR GARMENT FIDELITY
# ============================================================

import torch
import open_clip

print("=" * 70)
print("Q4 — LOADING OPENCLIP")
print("=" * 70)

device = "cuda" if torch.cuda.is_available() else "cpu"

CLIP_MODEL = "ViT-B-32"
CLIP_PRETRAINED = "laion2b_s34b_b79k"

clip_model, _, clip_preprocess = open_clip.create_model_and_transforms(
    CLIP_MODEL,
    pretrained=CLIP_PRETRAINED
)

clip_model = clip_model.to(device)
clip_model.eval()

print("Model      :", CLIP_MODEL)
print("Pretrained :", CLIP_PRETRAINED)
print("Device     :", device)

print()
print("✅ OpenCLIP loaded successfully")
print("=" * 70)
````

````python
# ============================================================
# Q4 — CELL 6
# GARMENT REGION + OPENCLIP FIDELITY
# ============================================================

import os
import cv2
import numpy as np
import torch
import torch.nn.functional as F

from PIL import Image

print("=" * 70)
print("Q4 — GARMENT FIDELITY METRIC")
print("=" * 70)

Q3_MASK_DIR = os.path.join(
    PROJECT_DIR,
    "Q3",
    "catvton_masks"
)


# ------------------------------------------------------------
# FIND FINAL Q3 MASK
# ------------------------------------------------------------

def find_q3_mask(pair_id):

    candidates = [
        f"{pair_id}_catvton_mask_final.png",
        f"{pair_id}_mask.png",
    ]

    for filename in candidates:

        path = os.path.join(
            Q3_MASK_DIR,
            filename
        )

        if os.path.isfile(path):
            return path

    return None


# ------------------------------------------------------------
# EXTRACT GENERATED GARMENT REGION
# ------------------------------------------------------------

def extract_garment_region(output_path, mask_path):
    """
    Extract the generated garment region from the Q3 output
    using the final person-space CatVTON mask.

    The region outside the mask is replaced with a neutral
    background before OpenCLIP embedding.
    """

    output = Image.open(output_path).convert("RGB")
    mask = Image.open(mask_path).convert("L")

    if mask.size != output.size:
        mask = mask.resize(
            output.size,
            Image.Resampling.NEAREST
        )

    output_np = np.array(output)
    mask_np = np.array(mask) > 127

    if not np.any(mask_np):
        raise ValueError(
            f"Empty garment mask: {mask_path}"
        )

    # Bounding box of garment/editable region
    ys, xs = np.where(mask_np)

    x1 = xs.min()
    x2 = xs.max()
    y1 = ys.min()
    y2 = ys.max()

    # Crop output and mask to relevant region
    crop = output_np[y1:y2 + 1, x1:x2 + 1]
    crop_mask = mask_np[y1:y2 + 1, x1:x2 + 1]

    # Neutral background outside garment region
    neutral = np.full_like(
        crop,
        255
    )

    garment_region = np.where(
        crop_mask[..., None],
        crop,
        neutral
    )

    return Image.fromarray(
        garment_region.astype(np.uint8)
    )


# ------------------------------------------------------------
# OPENCLIP EMBEDDING
# ------------------------------------------------------------

@torch.no_grad()
def clip_embedding(image):
    """
    Generate normalized OpenCLIP image embedding.
    """

    tensor = clip_preprocess(image).unsqueeze(0).to(device)

    features = clip_model.encode_image(tensor)

    features = F.normalize(
        features.float(),
        dim=-1
    )

    return features


# ------------------------------------------------------------
# GARMENT FIDELITY SCORE
# ------------------------------------------------------------

@torch.no_grad()
def garment_fidelity_score(
    garment_path,
    output_path,
    mask_path
):
    """
    Cosine similarity between:
      1. input garment image
      2. generated garment region from Q3 output

    Higher score = better garment fidelity.
    """

    garment = Image.open(
        garment_path
    ).convert("RGB")

    generated_region = extract_garment_region(
        output_path,
        mask_path
    )

    garment_embedding = clip_embedding(
        garment
    )

    output_embedding = clip_embedding(
        generated_region
    )

    similarity = F.cosine_similarity(
        garment_embedding,
        output_embedding
    ).item()

    return float(similarity)


# ------------------------------------------------------------
# VERIFY ALL FIVE MASKS
# ------------------------------------------------------------

all_masks_ready = True

for pair in pairs:

    mask_path = find_q3_mask(
        pair["pair_id"]
    )

    pair["mask_path"] = mask_path

    if mask_path is None:

        print(
            f"❌ {pair['pair_id']} -> mask NOT FOUND"
        )

        all_masks_ready = False

    else:

        print(
            f"✅ {pair['pair_id']} -> "
            f"{os.path.basename(mask_path)}"
        )


print()

if all_masks_ready:
    print("✅ Garment-region extraction ready")
    print("✅ OpenCLIP fidelity metric ready")
else:
    print("❌ One or more Q3 masks are missing")

print("=" * 70)
````

````python
# ============================================================
# Q4 — CELL 7
# IDENTITY PRESERVATION — FACE-REGION SSIM
# ============================================================

import cv2
import numpy as np

from skimage.metrics import structural_similarity as ssim

print("=" * 70)
print("Q4 — IDENTITY PRESERVATION METRIC")
print("=" * 70)

# ------------------------------------------------------------
# LOAD OPENCV FACE DETECTOR
# ------------------------------------------------------------

face_cascade_path = (
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

face_detector = cv2.CascadeClassifier(
    face_cascade_path
)

if face_detector.empty():
    raise RuntimeError(
        "Could not load OpenCV face detector."
    )

print("✅ OpenCV face detector loaded")


# ------------------------------------------------------------
# DETECT FACE
# ------------------------------------------------------------

def extract_face_region(
    image_path,
    padding=0.20
):
    """
    Detect the largest face in an image and return
    the padded RGB face crop.

    Returns:
        face_crop
        bbox = (x1, y1, x2, y2)

    If no face is detected:
        returns None, None
    """

    image = cv2.imread(image_path)

    if image is None:
        raise FileNotFoundError(
            f"Could not read image: {image_path}"
        )

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    faces = face_detector.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(50, 50)
    )

    if len(faces) == 0:
        return None, None

    # Select largest detected face
    x, y, w, h = max(
        faces,
        key=lambda box: box[2] * box[3]
    )

    pad_x = int(w * padding)
    pad_y = int(h * padding)

    x1 = max(0, x - pad_x)
    y1 = max(0, y - pad_y)

    x2 = min(
        image.shape[1],
        x + w + pad_x
    )

    y2 = min(
        image.shape[0],
        y + h + pad_y
    )

    face = image[
        y1:y2,
        x1:x2
    ]

    face = cv2.cvtColor(
        face,
        cv2.COLOR_BGR2RGB
    )

    return face, (x1, y1, x2, y2)


# ------------------------------------------------------------
# FACE SSIM
# ------------------------------------------------------------

def identity_preservation_score(
    person_path,
    output_path
):
    """
    Calculate SSIM between the detected face in the
    original person image and the detected face in the
    Q3 try-on output.

    Score range:
        1.0 = extremely similar
        lower = greater structural change
    """

    original_face, original_bbox = extract_face_region(
        person_path
    )

    output_face, output_bbox = extract_face_region(
        output_path
    )

    if original_face is None:
        raise RuntimeError(
            f"Face not detected in original person: {person_path}"
        )

    if output_face is None:
        raise RuntimeError(
            f"Face not detected in Q3 output: {output_path}"
        )

    # Standardize both face crops
    FACE_SIZE = (256, 256)

    original_face = cv2.resize(
        original_face,
        FACE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    output_face = cv2.resize(
        output_face,
        FACE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    # Convert to grayscale for structural comparison
    original_gray = cv2.cvtColor(
        original_face,
        cv2.COLOR_RGB2GRAY
    )

    output_gray = cv2.cvtColor(
        output_face,
        cv2.COLOR_RGB2GRAY
    )

    score = ssim(
        original_gray,
        output_gray,
        data_range=255
    )

    return float(score), original_bbox, output_bbox


print("✅ Face-region extraction ready")
print("✅ Face-region SSIM metric ready")
print("✅ Identity score range: 0–1")
print("=" * 70)
````

````python
# ============================================================
# Q4 — CELL 9
# FINAL IDENTITY METRIC — ALIGNED FACE-REGION SSIM
# ============================================================

import cv2
from skimage.metrics import structural_similarity as ssim

print("=" * 70)
print("Q4 — FINAL FACE-REGION SSIM")
print("=" * 70)


def identity_preservation_score(person_path, output_path):
    """
    Face-region SSIM.

    The face bbox is detected from the ORIGINAL person image
    and the same spatial region is extracted from both the
    original and generated image.

    This avoids inconsistent face detector boxes between
    original and output images.
    """

    original = cv2.imread(person_path)
    generated = cv2.imread(output_path)

    if original is None:
        raise FileNotFoundError(person_path)

    if generated is None:
        raise FileNotFoundError(output_path)

    # Generated Q3 outputs should correspond spatially to person
    if generated.shape[:2] != original.shape[:2]:
        generated = cv2.resize(
            generated,
            (original.shape[1], original.shape[0]),
            interpolation=cv2.INTER_AREA
        )

    # Detect face on ORIGINAL image
    _, bbox = extract_face_region(
        person_path,
        padding=0.20
    )

    if bbox is None:
        raise RuntimeError(
            f"Face not detected: {person_path}"
        )

    x1, y1, x2, y2 = [
        int(v) for v in bbox
    ]

    # Same spatial region from both images
    original_face = original[
        y1:y2,
        x1:x2
    ]

    generated_face = generated[
        y1:y2,
        x1:x2
    ]

    if original_face.size == 0 or generated_face.size == 0:
        raise RuntimeError(
            f"Invalid face crop for {person_path}"
        )

    # Standardize
    FACE_SIZE = (256, 256)

    original_face = cv2.resize(
        original_face,
        FACE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    generated_face = cv2.resize(
        generated_face,
        FACE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    # Grayscale SSIM
    original_gray = cv2.cvtColor(
        original_face,
        cv2.COLOR_BGR2GRAY
    )

    generated_gray = cv2.cvtColor(
        generated_face,
        cv2.COLOR_BGR2GRAY
    )

    score = ssim(
        original_gray,
        generated_gray,
        data_range=255
    )

    return float(score), bbox


print("✅ Original-person face bbox used for alignment")
print("✅ Same spatial face region compared in Q3 output")
print("✅ Metric: SSIM")
print("=" * 70)
````

````python
# ============================================================
# Q4 — CELL 10
# CALCULATE GARMENT FIDELITY + IDENTITY PRESERVATION
# ============================================================

import pandas as pd

print("=" * 70)
print("Q4 — CALCULATING AUTOMATED METRICS")
print("=" * 70)

metric_results = []

for pair in pairs:

    pair_id = pair["pair_id"]

    print()
    print(f"Evaluating {pair_id} ...")

    # --------------------------------------------------------
    # A. GARMENT FIDELITY
    # --------------------------------------------------------

    garment_score = garment_fidelity_score(
        pair["garment_path"],
        pair["output_path"],
        pair["mask_path"]
    )

    # --------------------------------------------------------
    # B. IDENTITY PRESERVATION
    # --------------------------------------------------------

    identity_score, face_bbox = identity_preservation_score(
        pair["person_path"],
        pair["output_path"]
    )

    result = {
        "pair_id": pair_id,

        "person_path": pair["person_path"],
        "garment_path": pair["garment_path"],
        "output_path": pair["output_path"],
        "mask_path": pair["mask_path"],

        "garment_fidelity": garment_score,
        "identity_preservation": identity_score,

        "face_bbox": tuple(
            int(v) for v in face_bbox
        )
    }

    metric_results.append(result)

    print(
        f"  Garment fidelity     : {garment_score:.4f}"
    )

    print(
        f"  Identity preservation: {identity_score:.4f}"
    )


# ------------------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------------------

metrics_df = pd.DataFrame(metric_results)

print()
print("=" * 70)
print("FIRST TWO Q4 METRICS COMPLETE")
print("=" * 70)

display(
    metrics_df[
        [
            "pair_id",
            "garment_fidelity",
            "identity_preservation"
        ]
    ].round(4)
)

print()
print("✅ Axis A: OpenCLIP garment-region similarity")
print("✅ Axis B: Face-region SSIM")
print("=" * 70)
````

````python
# ============================================================
# Q4 — TEST REAL VLM-AS-JUDGE ON PAIR 01
# Q1 MiniCPM-V-2_6-int4
# ============================================================

import json
import re
from PIL import Image

print("=" * 70)
print("Q4 — TESTING VLM-AS-JUDGE ON PAIR 01")
print("=" * 70)


def parse_judge_response(response):
    """
    Extract:
        score  -> integer 1-10
        reason -> string

    from MiniCPM's JSON response.
    """

    if not isinstance(response, str):
        response = str(response)

    text = response.strip()

    # Remove possible markdown fences
    text = text.replace("```json", "")
    text = text.replace("```", "")
    text = text.strip()

    # Try direct JSON first
    try:
        data = json.loads(text)

    except json.JSONDecodeError:

        # Extract first JSON object if model added extra text
        match = re.search(
            r'\{.*?\}',
            text,
            flags=re.DOTALL
        )

        if not match:
            raise ValueError(
                "Could not find JSON in VLM response:\n"
                + text
            )

        data = json.loads(
            match.group(0)
        )

    if "score" not in data:
        raise ValueError(
            f"VLM response missing 'score': {data}"
        )

    if "reason" not in data:
        raise ValueError(
            f"VLM response missing 'reason': {data}"
        )

    score = int(data["score"])
    reason = str(data["reason"]).strip()

    if not 1 <= score <= 10:
        raise ValueError(
            f"Invalid VLM score: {score}"
        )

    if not reason:
        raise ValueError(
            "VLM returned an empty reason."
        )

    return score, reason


def vlm_judge_pair(pair):
    """
    Evaluate one Q3 try-on result using the SAME Q1 MiniCPM-V model.

    Image order:
        1. Original person
        2. Target garment
        3. Q3 generated output
    """

    person = Image.open(
        pair["person_path"]
    ).convert("RGB")

    garment = Image.open(
        pair["garment_path"]
    ).convert("RGB")

    output = Image.open(
        pair["output_path"]
    ).convert("RGB")

    # MiniCPM multi-image conversation.
    # Each PIL image appears before the text rubric.
    msgs = [
        {
            "role": "user",
            "content": [
                person,
                garment,
                output,
                JUDGE_RUBRIC
            ]
        }
    ]

    response = vlm_model.chat(
        image=None,
        msgs=msgs,
        tokenizer=vlm_tokenizer,
        sampling=False,
        max_new_tokens=350
    )

    score, reason = parse_judge_response(
        response
    )

    return {
        "score": score,
        "reason": reason,
        "raw_response": str(response)
    }


# ------------------------------------------------------------
# TEST ONLY PAIR 01
# ------------------------------------------------------------

test_pair = pairs[0]

print("Pair   :", test_pair["pair_id"])
print(
    "Person :",
    test_pair["person_path"]
)
print(
    "Garment:",
    test_pair["garment_path"]
)
print(
    "Output :",
    test_pair["output_path"]
)

print()
print("Running Q1 MiniCPM-V judge...")
print()

judge_test = vlm_judge_pair(
    test_pair
)

print("=" * 70)
print("PAIR 01 — VLM JUDGE RESULT")
print("=" * 70)

print(
    "Score :",
    judge_test["score"],
    "/ 10"
)

print()
print(
    "Reason:",
    judge_test["reason"]
)

print()
print("Raw response:")
print(
    judge_test["raw_response"]
)

print("=" * 70)
print("✅ REAL Q1 VLM-AS-JUDGE TEST COMPLETE")
print("=" * 70)
````

````python
# ============================================================
# Q4 — RECOMPUTE AXIS A + AXIS B
# A: GARMENT FIDELITY — OPENCLIP
# B: IDENTITY PRESERVATION — FACE-REGION SSIM
# ============================================================

import os
import cv2
import torch
import numpy as np
import pandas as pd
import open_clip

from PIL import Image
from skimage.metrics import structural_similarity as ssim
import torch.nn.functional as F

print("=" * 70)
print("Q4 — COMPUTING AXIS A + AXIS B")
print("=" * 70)

device = "cuda" if torch.cuda.is_available() else "cpu"

# ------------------------------------------------------------
# 1. LOAD OPENCLIP
# ------------------------------------------------------------

CLIP_MODEL = "ViT-B-32"
CLIP_PRETRAINED = "laion2b_s34b_b79k"

print("\nLoading OpenCLIP...")

clip_model, _, clip_preprocess = open_clip.create_model_and_transforms(
    CLIP_MODEL,
    pretrained=CLIP_PRETRAINED
)

clip_model = clip_model.to(device)
clip_model.eval()

print("✅ OpenCLIP loaded:", CLIP_MODEL)


# ------------------------------------------------------------
# 2. Q3 MASK DIRECTORY
# ------------------------------------------------------------

Q3_MASK_DIR = os.path.join(
    PROJECT_DIR,
    "Q3",
    "catvton_masks"
)


def find_q3_mask(pair_id):

    candidates = [
        f"{pair_id}_catvton_mask_final.png",
        f"{pair_id}_mask.png",
    ]

    for filename in candidates:

        path = os.path.join(
            Q3_MASK_DIR,
            filename
        )

        if os.path.isfile(path):
            return path

    return None


# ------------------------------------------------------------
# 3. EXTRACT GENERATED GARMENT REGION
# ------------------------------------------------------------

def extract_garment_region(output_path, mask_path):

    output = Image.open(
        output_path
    ).convert("RGB")

    mask = Image.open(
        mask_path
    ).convert("L")

    if mask.size != output.size:
        mask = mask.resize(
            output.size,
            Image.Resampling.NEAREST
        )

    output_np = np.array(output)
    mask_np = np.array(mask) > 127

    if not np.any(mask_np):
        raise ValueError(
            f"Empty mask: {mask_path}"
        )

    ys, xs = np.where(mask_np)

    x1 = int(xs.min())
    x2 = int(xs.max())
    y1 = int(ys.min())
    y2 = int(ys.max())

    crop = output_np[
        y1:y2 + 1,
        x1:x2 + 1
    ]

    crop_mask = mask_np[
        y1:y2 + 1,
        x1:x2 + 1
    ]

    # White neutral background outside evaluated region
    neutral = np.full_like(
        crop,
        255
    )

    region = np.where(
        crop_mask[..., None],
        crop,
        neutral
    )

    return Image.fromarray(
        region.astype(np.uint8)
    )


# ------------------------------------------------------------
# 4. OPENCLIP EMBEDDING
# ------------------------------------------------------------

@torch.no_grad()
def clip_embedding(image):

    tensor = (
        clip_preprocess(image)
        .unsqueeze(0)
        .to(device)
    )

    features = clip_model.encode_image(
        tensor
    )

    return F.normalize(
        features.float(),
        dim=-1
    )


@torch.no_grad()
def garment_fidelity_score(
    garment_path,
    output_path,
    mask_path
):

    garment = Image.open(
        garment_path
    ).convert("RGB")

    output_garment = extract_garment_region(
        output_path,
        mask_path
    )

    garment_features = clip_embedding(
        garment
    )

    output_features = clip_embedding(
        output_garment
    )

    score = F.cosine_similarity(
        garment_features,
        output_features
    ).item()

    return float(score)


# ------------------------------------------------------------
# 5. FACE DETECTOR
# ------------------------------------------------------------

face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades
    + "haarcascade_frontalface_default.xml"
)

if face_detector.empty():
    raise RuntimeError(
        "Could not load OpenCV face detector."
    )


def detect_original_face_bbox(
    image_path,
    padding=0.20
):

    image = cv2.imread(
        image_path
    )

    if image is None:
        raise FileNotFoundError(
            image_path
        )

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    faces = face_detector.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(50, 50)
    )

    if len(faces) == 0:
        return None

    x, y, w, h = max(
        faces,
        key=lambda box:
        box[2] * box[3]
    )

    pad_x = int(w * padding)
    pad_y = int(h * padding)

    x1 = max(
        0,
        x - pad_x
    )

    y1 = max(
        0,
        y - pad_y
    )

    x2 = min(
        image.shape[1],
        x + w + pad_x
    )

    y2 = min(
        image.shape[0],
        y + h + pad_y
    )

    return (
        int(x1),
        int(y1),
        int(x2),
        int(y2)
    )


# ------------------------------------------------------------
# 6. FACE-REGION SSIM
# ------------------------------------------------------------

def identity_preservation_score(
    person_path,
    output_path
):

    original = cv2.imread(
        person_path
    )

    generated = cv2.imread(
        output_path
    )

    if original is None:
        raise FileNotFoundError(
            person_path
        )

    if generated is None:
        raise FileNotFoundError(
            output_path
        )

    # Ensure spatial alignment
    if generated.shape[:2] != original.shape[:2]:

        generated = cv2.resize(
            generated,
            (
                original.shape[1],
                original.shape[0]
            ),
            interpolation=cv2.INTER_AREA
        )

    bbox = detect_original_face_bbox(
        person_path
    )

    if bbox is None:
        raise RuntimeError(
            f"Face not detected: {person_path}"
        )

    x1, y1, x2, y2 = bbox

    original_face = original[
        y1:y2,
        x1:x2
    ]

    generated_face = generated[
        y1:y2,
        x1:x2
    ]

    FACE_SIZE = (256, 256)

    original_face = cv2.resize(
        original_face,
        FACE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    generated_face = cv2.resize(
        generated_face,
        FACE_SIZE,
        interpolation=cv2.INTER_AREA
    )

    original_gray = cv2.cvtColor(
        original_face,
        cv2.COLOR_BGR2GRAY
    )

    generated_gray = cv2.cvtColor(
        generated_face,
        cv2.COLOR_BGR2GRAY
    )

    score = ssim(
        original_gray,
        generated_gray,
        data_range=255
    )

    return float(score)


# ------------------------------------------------------------
# 7. EVALUATE ALL FIVE PAIRS
# ------------------------------------------------------------

automated_results = []

print("\nRunning automated metrics...\n")

for pair in pairs:

    pair_id = pair["pair_id"]

    print(f"Evaluating {pair_id} ...")

    mask_path = find_q3_mask(
        pair_id
    )

    if mask_path is None:
        raise FileNotFoundError(
            f"Q3 final mask missing for {pair_id}"
        )

    garment_score = garment_fidelity_score(
        pair["garment_path"],
        pair["output_path"],
        mask_path
    )

    identity_score = identity_preservation_score(
        pair["person_path"],
        pair["output_path"]
    )

    automated_results.append({
        "pair_id": pair_id,
        "garment_fidelity": garment_score,
        "identity_preservation": identity_score
    })

    print(
        f"  Garment fidelity : {garment_score:.4f}"
    )

    print(
        f"  Face SSIM        : {identity_score:.4f}"
    )


# ------------------------------------------------------------
# 8. DATAFRAME
# ------------------------------------------------------------

automated_df = pd.DataFrame(
    automated_results
)

print()
print("=" * 70)
print("Q4 — AUTOMATED METRICS COMPLETE")
print("=" * 70)

display(
    automated_df.round(4)
)

print()
print("✅ Axis A — Garment fidelity complete")
print("✅ Axis B — Identity preservation complete")
print("=" * 70)
````

````python
# ============================================================
# Q4 — FILL OFFICIAL evaluation_template_q4.csv
# ============================================================

import os
import pandas as pd
import numpy as np

print("=" * 70)
print("Q4 — BUILDING FINAL EVALUATION CSV")
print("=" * 70)

# ------------------------------------------------------------
# 1. LOAD SAVED VLM RESULTS
# ------------------------------------------------------------

VLM_BACKUP_PATH = os.path.join(
    Q4_DIR,
    "vlm_judge_results.csv"
)

vlm_final_df = pd.read_csv(
    VLM_BACKUP_PATH
)

print("✅ VLM results loaded")


# ------------------------------------------------------------
# 2. MERGE ALL THREE EVALUATION AXES
# ------------------------------------------------------------

scores_df = automated_df.merge(
    vlm_final_df,
    on="pair_id",
    how="inner"
)

if len(scores_df) != 5:
    raise RuntimeError(
        f"Expected 5 evaluated pairs, found {len(scores_df)}"
    )


# ------------------------------------------------------------
# 3. LOAD FRESH COPY OF OFFICIAL TEMPLATE
# ------------------------------------------------------------

final_df = pd.read_csv(
    TEMPLATE_PATH
)

expected_columns = [
    "pair_id",
    "person_image",
    "garment_image",
    "tryon_model",
    "garment_fidelity_score",
    "identity_preservation_score",
    "vlm_judge_score",
    "vlm_judge_reasons",
    "artifacts_observed",
    "notes"
]

if list(final_df.columns) != expected_columns:
    raise RuntimeError(
        "Template columns do not match expected Q4 schema."
    )


# ------------------------------------------------------------
# 4. MAP ACTUAL FILE NAMES FROM OUR FIVE PAIRS
# ------------------------------------------------------------

pair_lookup = {
    pair["pair_id"]: pair
    for pair in pairs
}


# ------------------------------------------------------------
# 5. HELPER — SUMMARIZE ARTIFACTS FROM VLM REASON
# ------------------------------------------------------------

def artifact_summary(reason):

    text = str(reason).lower()

    # We are only summarizing what the VLM itself reported.
    if (
        "no visible artifacts" in text
        or "no visible artifact" in text
    ):
        return "No major visible artifacts reported by VLM judge."

    if "waist" in text:
        return "Minor garment boundary/fit issue around waist reported by VLM judge."

    if "minor imperfection" in text:
        return "Minor visual imperfections reported by VLM judge."

    return "See VLM judge reasons for observed quality issues."


# ------------------------------------------------------------
# 6. FILL EVERY TEMPLATE ROW
# ------------------------------------------------------------

for idx, row in final_df.iterrows():

    pair_id = row["pair_id"]

    if pair_id not in pair_lookup:
        raise RuntimeError(
            f"Unknown pair in template: {pair_id}"
        )

    pair = pair_lookup[pair_id]

    score_row = scores_df[
        scores_df["pair_id"] == pair_id
    ]

    if len(score_row) != 1:
        raise RuntimeError(
            f"Missing/duplicate metrics for {pair_id}"
        )

    score_row = score_row.iloc[0]

    # Actual filenames — replaces <candidate_sourced>
    final_df.at[idx, "person_image"] = os.path.basename(
        pair["person_path"]
    )

    final_df.at[idx, "garment_image"] = os.path.basename(
        pair["garment_path"]
    )

    # Q3 model
    final_df.at[idx, "tryon_model"] = "CatVTON"

    # Axis A
    final_df.at[
        idx,
        "garment_fidelity_score"
    ] = round(
        float(score_row["garment_fidelity"]),
        4
    )

    # Axis B
    final_df.at[
        idx,
        "identity_preservation_score"
    ] = round(
        float(score_row["identity_preservation"]),
        4
    )

    # Axis C
    final_df.at[
        idx,
        "vlm_judge_score"
    ] = int(
        score_row["vlm_score"]
    )

    reason = str(
        score_row["vlm_reason"]
    ).strip()

    final_df.at[
        idx,
        "vlm_judge_reasons"
    ] = reason

    final_df.at[
        idx,
        "artifacts_observed"
    ] = artifact_summary(reason)

    # Methodological note, NOT a fabricated evaluation score
    final_df.at[
        idx,
        "notes"
    ] = (
        "Garment fidelity: OpenCLIP ViT-B-32 cosine similarity "
        "on Q3 garment region; identity preservation: "
        "face-region SSIM; VLM judge: Q1 MiniCPM-V-2_6-int4."
    )


# ------------------------------------------------------------
# 7. VALIDATE FINAL DATA
# ------------------------------------------------------------

required_score_columns = [
    "garment_fidelity_score",
    "identity_preservation_score",
    "vlm_judge_score",
    "vlm_judge_reasons"
]

if final_df[required_score_columns].isna().any().any():
    raise RuntimeError(
        "Final Q4 CSV contains missing evaluation values."
    )

if not final_df["vlm_judge_score"].between(
    1,
    10
).all():
    raise RuntimeError(
        "Invalid VLM score outside 1–10."
    )

if len(final_df) != 5:
    raise RuntimeError(
        "Final Q4 CSV must contain exactly 5 pairs."
    )


# ------------------------------------------------------------
# 8. SAVE USING REQUIRED FILE NAME
# ------------------------------------------------------------

FINAL_CSV_PATH = os.path.join(
    Q4_DIR,
    "evaluation_template_q4.csv"
)

final_df.to_csv(
    FINAL_CSV_PATH,
    index=False
)


# ------------------------------------------------------------
# 9. DISPLAY FINAL SUBMISSION CSV
# ------------------------------------------------------------

print()
print("=" * 70)
print("Q4 — FINAL CSV")
print("=" * 70)

display(final_df)

print()
print("Saved:")
print(FINAL_CSV_PATH)

print()
print("Rows   :", len(final_df))
print("Columns:", len(final_df.columns))

print()
print("✅ 5 / 5 pairs evaluated")
print("✅ Official 10-column template preserved")
print("✅ OpenCLIP garment fidelity included")
print("✅ Face-region SSIM included")
print("✅ Q1 MiniCPM-V judge score included")
print("✅ Q1 MiniCPM-V judge reasons included")
print("✅ evaluation_template_q4.csv COMPLETE")
print("=" * 70)
````
